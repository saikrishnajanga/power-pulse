```python

# STEP 4: Preprocess data for solar generation prediction

import pandas as pd
import numpy as np

# Use only relevant prediction features; avoid leakage from
# consumption, grid import/export, and electricity cost.
feature_cols = [
    "timestamp",
    "city",
    "panel_capacity_kw",
    "battery_installed",
    "occupants",
    "irradiance_wm2",
    "temperature_c",
    "cloud_cover_pct"
]
target_col = "solar_generation_kwh"

# Convert timestamps
ts = pd.to_datetime(df["timestamp"], errors="coerce")

# Exclude invalid timestamps and missing target values
valid = ts.notna() & df[target_col].notna()
valid_idx = np.flatnonzero(valid.to_numpy())
valid_ts = ts.iloc[valid_idx]

# Chronological split: first 80% of time for training,
# latest 20% for testing
cutoff = valid_ts.quantile(0.80)

train_idx = valid_idx[(valid_ts < cutoff).to_numpy()]
test_idx = valid_idx[(valid_ts >= cutoff).to_numpy()]

# Use a reproducible sample for the first training run
rng = np.random.default_rng(42)
if len(train_idx) > 500_000:
    train_idx = rng.choice(train_idx, 500_000, replace=False)
if len(test_idx) > 100_000:
    test_idx = rng.choice(test_idx, 100_000, replace=False)

# Select required columns only
train_df = df.iloc[train_idx][feature_cols + [target_col]].copy()
test_df = df.iloc[test_idx][feature_cols + [target_col]].copy()

# Extract time-based features
for part in (train_df, test_df):
    part["timestamp"] = pd.to_datetime(
        part["timestamp"], errors="coerce"
    )
    part["hour"] = part["timestamp"].dt.hour
    part["day_of_year"] = part["timestamp"].dt.dayofyear
    part["month"] = part["timestamp"].dt.month
    part.drop(columns=["timestamp"], inplace=True)

# Separate features and target
X_train = train_df.drop(columns=[target_col])
y_train = train_df[target_col].astype("float32")

X_test = test_df.drop(columns=[target_col])
y_test = test_df[target_col].astype("float32")

# One-hot encode city without assigning artificial numeric order
X_train = pd.get_dummies(X_train, columns=["city"], dtype=np.uint8)
X_test = pd.get_dummies(X_test, columns=["city"], dtype=np.uint8)

# Ensure training and testing have identical feature columns
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

# Handle missing and infinite numeric values using training medians
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)

medians = X_train.median()
X_train = X_train.fillna(medians).astype("float32")
X_test = X_test.fillna(medians).astype("float32")

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training period ends before:", cutoff)
print("Missing values remaining in training:", int(X_train.isna().sum().sum()))
print("Missing values remaining in testing:", int(X_test.isna().sum().sum()))
print("Preprocessing completed!")
```

```python

# STEP 5: Train the solar energy prediction model

from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import joblib
import time

model = HistGradientBoostingRegressor(
    max_iter=100,
    learning_rate=0.1,
    max_leaf_nodes=31,
    l2_regularization=1.0,
    early_stopping=True,
    random_state=42,
    verbose=1
)

print("Starting model training...")
start_time = time.time()

model.fit(X_train, y_train)

training_time = time.time() - start_time
print(f"\nTraining completed in {training_time:.2f} seconds")

# Predict on the held-out, later time period
y_pred = model.predict(X_test)

# Evaluate performance
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\n--- SOLAR GENERATION MODEL RESULTS ---")
print(f"MAE:  {mae:.4f} kWh")
print(f"RMSE: {rmse:.4f} kWh")
print(f"R²:   {r2:.4f}")
print(f"Training samples: {len(X_train):,}")
print(f"Testing samples:  {len(X_test):,}")

# Save model and its expected feature order to Google Drive
model_path = (
    "/content/drive/MyDrive/solar_energy/"
    "solar_generation_model.joblib"
)

joblib.dump(
    {"model": model, "feature_columns": X_train.columns.tolist(),
     "medians": medians},
    model_path
)

print("\nSaved trained model to:", model_path)
```

```python
import ipywidgets as widgets
import pandas as pd
from IPython.display import display, clear_output

city_cols = [c for c in X_train.columns if c.startswith("city_")]

cap   = widgets.FloatText(value=5.0, description="Panel kW")
bat   = widgets.Dropdown(options=[("No", 0), ("Yes", 1)], description="Battery")
occ   = widgets.IntSlider(value=3, min=1, max=6, description="Occupants")
irr   = widgets.FloatSlider(value=600, min=0, max=1100, description="Irradiance")
temp  = widgets.FloatSlider(value=30, min=5, max=46, description="Temp °C")
cloud = widgets.FloatSlider(value=25, min=0, max=100, description="Cloud %")
hour  = widgets.IntSlider(value=12, min=0, max=23, description="Hour")
date  = widgets.DatePicker(value=pd.Timestamp("2025-11-15").date(), description="Date")
city  = widgets.Dropdown(options=[c[5:] for c in city_cols], description="City")
btn   = widgets.Button(description="Predict", button_style="success")
out   = widgets.Output()

def predict(_):
    dt = pd.Timestamp(date.value)
    row = pd.DataFrame([medians]).astype("float32")
    row[city_cols] = 0
    row["city_" + city.value] = 1
    row["panel_capacity_kw"] = cap.value
    row["battery_installed"] = bat.value
    row["occupants"] = occ.value
    row["irradiance_wm2"] = irr.value
    row["temperature_c"] = temp.value
    row["cloud_cover_pct"] = cloud.value
    row["hour"] = hour.value
    row["day_of_year"] = dt.dayofyear
    row["month"] = dt.month
    p = max(0.0, float(model.predict(row[X_train.columns])[0]))
    with out:
        clear_output()
        print(f"Predicted solar generation: {p:.3f} kWh")

btn.on_click(predict)
display(cap, bat, occ, irr, temp, cloud, hour, date, city, btn, out)
```
